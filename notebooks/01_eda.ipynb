# Stroke Risk - Exploratory Data Analysis
Dataset: Kaggle Stroke Prediction Dataset (fedesoriano).

```python
import sys; sys.path.insert(0,'../src')
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from stroke_risk.data import load_raw, validate, clean
df = load_raw()
print(validate(df))
df = clean(df)
df.head()
```

```python
print('Stroke rate:', df.stroke.mean())
sns.countplot(x='stroke', data=df); plt.show()
sns.histplot(df.age, kde=True); plt.title('Age distribution'); plt.show()
```
